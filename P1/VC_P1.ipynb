```python
vid = cv2.VideoCapture(0)

# Tipografía para mostrar texto
font = cv2.FONT_HERSHEY_SIMPLEX
  
lanzado = 0 #bandera para primer fotograma
px = -1
while(True):      
    # fotograma a fotograma
    ret, frame = vid.read()

    if ret: 
        # Activa el manejador en el primer fotograma
        if lanzado == 0:
            # Muestra fotograma
            cv2.imshow('Cam', frame)   
            # Define el nombre del manejador del evento
            cv2.setMouseCallback('Cam', mouse_events)
            lanzado = 1
            
        # Muestra valores RGB intentando centrar en el puntero
        if px > -1:
            cv2.putText(frame, '{}'.format(R), (px-45,py-5), font, 0.5, (0, 0, 255), 1)
            cv2.putText(frame, '     {}'.format(G), (px-54,py-5), font, 0.5, (0, 255, 0), 1)
            cv2.putText(frame, '         {}'.format(B), (px-54,py-5), font, 0.5, (255, 0, 0), 1)
            
        cv2.imshow('Cam', frame)   
        
    # Detenemos pulsado ESC
    if cv2.waitKey(20) == 27:
        break
  
# Libera el objeto de captura
vid.release()
# Destruye ventanas
cv2.destroyAllWindows()
```

Pop art
Salida de la webcam "inspirada" en Andy Warhol (Marilyn, https://temasycomentariosartepaeg.blogspot.com/p/autor-andy-warhol-1928-1987-titulo.html)

```python
import cv2

# Encendemos la cámara
vid = cv2.VideoCapture(0)

while(True):
    ret, frame = vid.read()
    if not ret:
        break

    # La paso a gris
    frame_gris = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    
    alto = frame_gris.shape[0]
    ancho = frame_gris.shape[1]
    
    valor_minimo = 255 
    valor_maximo = 0
    coordenada_min = (0, 0)
    coordenada_max = (0, 0)
    
    # Comprobamos en todo el frame cada pixel
    for y in range(alto):
        for x in range(ancho):
            pixel = frame_gris[y, x]
            
            if pixel < valor_minimo:
                valor_minimo = pixel
                coordenada_min = (x, y)
                
            if pixel > valor_maximo:
                valor_maximo = pixel
                coordenada_max = (x, y)
                
    # El más oscuro en rojo
    cv2.circle(frame, coordenada_min, 30, (0, 0, 255), 2)
    # El más claro en azul
    cv2.circle(frame, coordenada_max, 30, (255, 0, 0), 2)
    
    # Muestro el resultado
    cv2.imshow('Cam', frame)
    
    # Tecla ESC para salir
    if cv2.waitKey(20) == 27:
        break

vid.release()
cv2.destroyAllWindows()
```

Transformando la imagen capturada en una colección de círculos, cuyo radio depende de la intensidad promedio de la zona

```python
vid = cv2.VideoCapture(0)

# Se define el número de celdas
ncells = 10
off = int(ncells/2)
  
while(True):      
    # fotograma a fotograma
    ret, frame = vid.read()

    if ret:
        #Dimensiones originales
        h, w, c = frame.shape
        #Redimensiono teniendo en cuenta ncells
        down_frame = cv2.resize(frame, (int(w/ncells),int(h/ncells)),cv2.INTER_NEAREST)
        #Dimensiones reducidas
        h2, w2, c2 = down_frame.shape

        #Separamos canales
        r = down_frame[:,:,0]
        g = down_frame[:,:,1]
        b = down_frame[:,:,2]

        # Creamos imagen negra
        gris_up_frame = np.zeros((h2*ncells,w2*ncells,1), dtype = np.uint8)

        # recorre valores de la imagen reducida
        for y in range(0,h2):
            for x in range(0,w2):
                # La suma de los valores RGB define el radio del círculo
                rad = int ( (r[y,x] + g[y,x] + b[y,x])/(ncells*3*2))
                cv2.circle(gris_up_frame,(x*ncells+off, y*ncells + off), rad, (255,255,255), -1)

    
        # Muestra fotograma resultante
        cv2.imshow('Cam', gris_up_frame)
    
    # Detenemos pulsado ESC
    if cv2.waitKey(20) == 27:
        break
  
# Libera el objeto de captura
vid.release()
# Destruye ventanas
cv2.destroyAllWindows()
```

Hemos concluido que vamos a crear un filtro que en escala de grises, los colores por debajo de 30 de valor de gris van a tener cierto rango de un color al azar, por debajo de 60, otro color, y así hasta 255. Logrando que las zonas oscuras tengan un tono con un color que nosotros decidamos, y las claras con otro. El resultado sería que las zonas oscuras de la imagen tienen un color, las más claras otro y las blancas otro. Creando una reconstrución de estilo "Pop Art"

Este fue el prompt mandado a la IA

```python
import cv2
import numpy as np

# 1. Definir la paleta Pop Art (LUT de 256 valores y 3 canales BGR)
lut = np.zeros((256, 3), dtype=np.uint8)

# Asignamos colores estridentes según la luminosidad (Azul, Verde, Rojo)
for i in range(256):
    if i < 30:
        lut[i] = [130, 0, 75]       # Morado oscuro (sombras profundas)
    elif i < 80:
        lut[i] = [255, 0, 0]        # Azul eléctrico
    elif i < 140:
        lut[i] = [0, 255, 0]        # Verde radiactivo
    elif i < 10:
        lut[i] = [0, 255, 255]      # Amarillo puro
    elif i < 225:
        lut[i] = [0, 100, 255]      # Naranja flúor
    else:
        lut[i] = [204, 50, 255]     # Rosa chicle (luces más altas)

# 2. Iniciar la captura de vídeo
vid = cv2.VideoCapture(0)

while True:
    ret, frame = vid.read()
    if not ret:
        break
    
    # Obtenemos el mapa de luces y sombras del fotograma
    frame_gris = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # 3. Mapeo de color instantáneo
    # NumPy busca el valor de cada píxel gris (0-255) y lo sustituye por el color de la LUT
    frame_popart = lut[frame_gris]

    # Mostramos el resultado
    cv2.imshow('Pop Art - Fleje Colores', frame_popart)

    # Detenemos al pulsar ESC
    if cv2.waitKey(20) == 27:
        break

vid.release()
cv2.destroyAllWindows()
```
